```python
df = pd.read_csv("netflix_titles.csv")

print("Dataset shape:", df.shape)

df.head()
```

```python
print("Column Names:")
print(df.columns.tolist())

print("\nData Types:")
print(df.dtypes)

print("\nMissing Values:")
print(df.isnull().sum())

print("\nDuplicate Rows:")
print(df.duplicated().sum())
```

```python
df.columns = (
    df.columns
    .str.strip()
    .str.lower()
    .str.replace(" ", "_")
)

print("Cleaned column names:")
print(df.columns.tolist())
```

```python
print("Missing values before cleaning:")
print(df.isnull().sum()[df.isnull().sum() > 0])

# Fill missing text values with a clear label
df["director"] = df["director"].fillna("Not Available")
df["cast"] = df["cast"].fillna("Not Available")
df["country"] = df["country"].fillna("Not Available")
df["rating"] = df["rating"].fillna("Not Rated")

# Convert date_added to date type
df["date_added"] = pd.to_datetime(
    df["date_added"],
    errors="coerce"
)

# Fill missing dates with a clear label later after date formatting
print("\nMissing values after initial cleaning:")
print(df.isnull().sum()[df.isnull().sum() > 0])
```

```python
duplicates_before = df.duplicated().sum()

df = df.drop_duplicates().reset_index(drop=True)

print("Duplicates before:", duplicates_before)
print("Duplicates after:", df.duplicated().sum())
```

```python
text_columns = [
    "type",
    "title",
    "director",
    "cast",
    "country",
    "rating",
    "duration",
    "listed_in",
    "description"
]

for column in text_columns:
    if column in df.columns:
        df[column] = df[column].astype(str).str.strip()

# Standardize selected categorical fields
df["type"] = df["type"].str.title()
df["country"] = df["country"].str.title()
df["rating"] = df["rating"].str.upper()

print("Type values:")
print(df["type"].unique())

print("\nRating values:")
print(df["rating"].unique())

# Convert date_added to a proper datetime type
df["date_added"] = pd.to_datetime(
    df["date_added"],
    errors="coerce"
)

# Replace missing dates with a clear label for the final CSV
df["date_added"] = df["date_added"].dt.strftime("%d-%m-%Y")
df["date_added"] = df["date_added"].fillna("Not Available")

print(df["date_added"].head())
print("Missing date_added values:", df["date_added"].isnull().sum())

df["release_year"] = pd.to_numeric(
    df["release_year"],
    errors="coerce"
).astype("Int64")

print(df.dtypes)

print("Final dataset shape:", df.shape)

print("\nMissing values after cleaning:")
print(df.isnull().sum())

print("\nDuplicate rows after cleaning:")
print(df.duplicated().sum())

print("\nFinal data types:")
print(df.dtypes)

df.head()
df.to_csv("cleaned_netflix_titles.csv", index=False)

from google.colab import files
files.download("cleaned_netflix_titles.csv")
```
